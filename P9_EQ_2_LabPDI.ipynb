### **EJERCICIO 1.** ***Ecualización del valor “value” en la representación de colores HSV.***

Para la realización de esta fase de la actividad utilizaremos la imagen de un Glaucoma proporcionado por el profesor.
glaucoma.jpg

Utilizaremos diversas librerías que nos permitirán realizar nuestras operaciones, pero la principal es **OpenCV**. OpenCV es una librería que permite procesar imágenes y extraer características usando Python.
También, utilizaremos **Matplotlib**, esta se emplea para graficar y visualizar la imagen de gráficos en un panel de visualización

```python
import cv2
import matplotlib.pyplot as plt
```

Cargaremos la imagen médica con el comando cv2.imread, y para el preprocesamiento de la imagen, leeremos la imagen en BGR por medio de OpenCV, con ello, convertiremos la imagen de un espacio de color a otro, BGR → RGB. Se realizará mediante el comando ***cv2.COLOR_BGR2RGB***.

```python
img_bgr = cv2.imread("glaucoma.jpg", cv2.COLOR_BGR2RGB)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
```

Después de cargar y convertir la imagen, utilizaremos la imagen original en formato BGR para cambiarla a otro espacio de color, el HSV, el cual significa *Hue* (tono o color), *Saturation* (intensidad de color) y *Value* (brillo). Se realizará mediante el comando ***cv2.COLOR_BGR2HSV.***


```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)
```

Posteriormente, separaremos los canales de nuestra imagen HSV, permitiéndonos dividir la imagen en tres matrices, una para el tono o color, otra para la saturación y la última para el brillo, esto mediante la función ***cv2.split()***.

```python
h, s, v = cv2.split(img_hsv)
```

Elegiremos el canal del *valor del brillo o iluminación*, esto lo hacemos para poder realizar modificaciones en este mismo sin alterar los colores de los otros canales. Realizaremos una *ecualización* del canal, la cual nos permite redistribuir los valores de intensidad de la imagen, además de aumentar el contraste, lo realizamos mediante la función ***cv2.equalizeHist()***.

```python
v_eq = cv2.equalizeHist(v)
```

Después de realizar la ecualización del canal *“Value”* y mejorar sus características, realizaremos una unión nuevamente de los canales previamente separados. Se unirán los canales en una nueva imagen, esto mediante la función ***cv2.merge()***.

```python
img_hsv_eq = cv2.merge([h, s, v_eq])
```

Seguidamente, después de unir los canales y obtener nuestra imagen HSV con brillo mejorado, convertiremos la nueva imagen HSV de nuevo a BGR y en seguida a BGR para su visualización, esto mediante las funciones ***cv2.COLOR_HSV2BGR*** y ***cv2.COLOR_BGR2RGB***.

```python
img_eq_bgr = cv2.cvtColor(img_hsv_eq, cv2.COLOR_HSV2BGR)
img_eq_rgb = cv2.cvtColor(img_eq_bgr, cv2.COLOR_BGR2RGB)
```

Mediante el código realizado se obtiene lo siguiente:

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb)
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_eq_rgb)
plt.title('Imagen mejorada (Ecualización V)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

Al aplicar la ecualización del histograma de la imagen HSV en el canal “Value” se mejora el contraste de la imagen, en lo cual podemos observar, el realce de las estructuras de los vasos sanguíneos. Sin embargo, también se observa un aumento de ruido en algunas regiones debido a que la ecualización global puede aumentar diferentes intensidades en toda la imagen.

### **EJERCICIO 2.** ***Ecualización del valor “value” en la representación de colores HSV en dos imagenes medicas diferentes.***

Al aplicar el código explicado en otras imágenes, tal como: *Resonancia Magnética* e imagen de *Histología*, obtenemos lo siguiente:

```python
img_bgr1 = cv2.imread("RMcerebral.jpg", cv2.COLOR_BGR2RGB)
img_rgb1 = cv2.cvtColor(img_bgr1, cv2.COLOR_BGR2RGB)

img_hsv1 = cv2.cvtColor(img_bgr1, cv2.COLOR_BGR2HSV)

h1, s1, v1 = cv2.split(img_hsv1)

v_eq1 = cv2.equalizeHist(v1)

img_hsv_eq1 = cv2.merge([h1, s1, v_eq1])

img_eq_bgr1 = cv2.cvtColor(img_hsv_eq1, cv2.COLOR_HSV2BGR)
img_eq_rgb1 = cv2.cvtColor(img_eq_bgr1, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb1)
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_eq_rgb1)
plt.title('Imagen mejorada (Ecualización V)')
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
img_bgr2 = cv2.imread("imagen_histologia.jpeg", cv2.COLOR_BGR2RGB)
img_rgb2 = cv2.cvtColor(img_bgr2, cv2.COLOR_BGR2RGB)

img_hsv2 = cv2.cvtColor(img_bgr2, cv2.COLOR_BGR2HSV)

h2, s2, v2 = cv2.split(img_hsv2)

v_eq2 = cv2.equalizeHist(v2)

img_hsv_eq2 = cv2.merge([h2, s2, v_eq2])

img_eq_bgr2 = cv2.cvtColor(img_hsv_eq2, cv2.COLOR_HSV2BGR)
img_eq_rgb2 = cv2.cvtColor(img_eq_bgr2, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_rgb2)
plt.title('Imagen original')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(img_eq_rgb2)
plt.title('Imagen mejorada (Ecualización V)')
plt.axis('off')

plt.tight_layout()
plt.show()
```
